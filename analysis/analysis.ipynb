```python
## Number of Employee vs Department
department_count= df["Department"].value_counts()

plt.figure(figsize=(8,5))
department_count.plot(kind="bar")
plt.title("Employee by dept-bar chart")
plt.xlabel("Department")
plt.ylabel("Number of employee")
plt.xticks(rotation=0)
plt.show()
```

```python
job_role_count= df["JobRole"].value_counts()

plt.figure(figsize=(15,10))
job_role_count.plot(kind="bar")
plt.title("Employee By Job role")
plt.ylabel("Mumber of Employee")
plt.xlabel("Job Role")
plt.xticks(rotation=20, ha="right")
plt.show()
```

```python
#employee leaving the company
attrition_count=df["Attrition"].value_counts()

plt.figure(figsize=(6,6))
attrition_count.plot(kind="pie",autopct="%1.1f%%")
plt.title("Employee attrition distribution")
plt.show()
```

```python
attrition_dept= pd.crosstab(#used to link 2 or more dept
    df["Department"],
    df["Attrition"]
)

plt.figure(figsize=(8,5))
attrition_dept.plot(kind="bar")
plt.title("Attrition by Dept.")
plt.xlabel("Department")
plt.ylabel("Number of employee")
plt.xticks(rotation=0)
plt.legend(title="Attrition")
plt.show()
```

```python
overtime_attrition=pd.crosstab(
    df["OverTime"],
    df["Attrition"]
)

plt.figure(figsize=(7,5))
overtime_attrition.plot(kind="bar")
plt.title("Overtime vs Attrition")
plt.xlabel("Overtime")
plt.ylabel("Attrition")
plt.xticks(rotation=0)
plt.legend(title="Attrition")
plt.show(
    
)
```

```python
income_level=df.groupby(
    "JobLevel"
)["MonthlyIncome"].mean()

plt.figure(figsize=(8,5))
income_level.plot(kind="bar")
plt.title("Average Monthly Income by Job Level")
plt.xlabel("Job Level")
plt.ylabel("Average Monthly Income")
plt.xticks(rotation=0)
plt.show()
```

```python
job_satisfaction=df["JobSatisfaction"].value_counts().sort_index() #sorts based on job satis from 1 to 4

plt.figure(figsize=(7,5))
job_satisfaction.plot(kind="bar")
plt.title("Job Satisfaction Distribution")
plt.xlabel("Job Satisfaction Level")
plt.ylabel("Number of Employees")
plt.xticks(rotation=0)
plt.show()
```

```python
plt.figure(figsize=(8,5))
plt.hist(df["Age"], bins=15)
plt.title("Employee Age Distribution")
plt.xlabel("Age")
plt.ylabel("Number of Employees")
plt.show()
```

```python
import seaborn as sns

numeric_df = df.select_dtypes(include= "number")#select only numbers

plt.figure(figsize=(14,10))

sns.heatmap(
    numeric_df.corr(),#tells to provide correlation 
    annot=False,
    cmap="coolwarm"
    
    #negative correlation → cooler colors
    #positive correlation → warmer colors
)
plt.xticks(rotation=90)

plt.title("HR Dataset Correlation Heatmap")
plt.show()
```
